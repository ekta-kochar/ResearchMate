```python
text_columns = [
    "query",
    "source",
    "source_id",
    "title",
    "abstract",
    "doi",
    "url"
]

for col in text_columns:
    if col in df.columns:
        df[col] = df[col].apply(
            lambda x: x.strip()
            if isinstance(x, str)
            else x
        )
```

```python
def normalize_doi(doi):

    if pd.isna(doi):
        return None

    doi = str(doi).strip().lower()

    doi = doi.replace(
        "https://doi.org/",
        ""
    )

    doi = doi.replace(
        "http://doi.org/",
        ""
    )

    doi = doi.replace(
        "doi:",
        ""
    )

    doi = doi.strip()

    return doi if doi else None
```

```python
df["doi_normalized"] = df["doi"].apply(
    normalize_doi
)
```

```python
df[
    ["doi", "doi_normalized"]
].head(10)
```

```python
import re


def normalize_title(title):

    if pd.isna(title):
        return None

    title = str(title).lower().strip()

    title = re.sub(
        r"[^\w\s]",
        "",
        title
    )

    title = re.sub(
        r"\s+",
        " ",
        title
    )

    return title
```

```python
df["title_normalized"] = df["title"].apply(
    normalize_title
)
```

```python
df[
    ["title", "title_normalized"]
].head()
```

```python
# DOI duplicates
doi_duplicates = df[
    df["doi_normalized"].notna()
    & df["doi_normalized"].duplicated(keep=False)
]

print(
    "Rows involved in DOI duplicates:",
    len(doi_duplicates)
)
```

```python
no_doi = df[
    df["doi_normalized"].isna()
].copy()

title_duplicates_no_doi = no_doi[
    no_doi["title_normalized"].notna()
    & no_doi["title_normalized"].duplicated(keep=False)
]

print(
    "Rows involved in title duplicates without DOI:",
    len(title_duplicates_no_doi)
)
```

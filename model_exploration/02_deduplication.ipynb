```python
import pandas as pd
```

```python
df = pd.read_csv(
    "outputs/cleaned_papers.csv"
)

df.head()
```

```python
with_doi = df[
    df["doi_normalized"].notna()
].copy()

without_doi = df[
    df["doi_normalized"].isna()
].copy()
```

```python
with_doi_dedup = with_doi.drop_duplicates(
    subset=["doi_normalized"],
    keep="first"
)
```

```python
without_doi_dedup = without_doi.copy()
```

```python
df_dedup = pd.concat(
    [with_doi_dedup, without_doi_dedup],
    ignore_index=True
)
```

```python
print("Before:", len(df))
print("After:", len(df_dedup))
print("Removed:", len(df) - len(df_dedup))
```

```python
df_dedup.to_csv(
    "outputs/deduplicated_papers.csv",
    index=False
)
```

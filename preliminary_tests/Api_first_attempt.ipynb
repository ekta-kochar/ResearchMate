```python
import requests

url = "https://www.ebi.ac.uk/europepmc/webservices/rest/search"

params = {
    "query": "metformin cancer",
    "format": "json",
    "resultType": "core",
    "pageSize": 5
}

response = requests.get(url, params=params)

data = response.json()

papers = data["resultList"]["result"]


def normalize_europe_pmc_paper(paper):
    return {
        "paper_id": paper.get("id"),
        "title": paper.get("title"),
        "authors": paper.get("authorString"),
        "abstract": paper.get("abstractText"),
        "year": paper.get("pubYear"),
        "doi": paper.get("doi"),
        "journal": paper.get("journalTitle"),
        "source": "Europe PMC",
        "source_id": paper.get("id")
    }


normalized_papers = [
    normalize_europe_pmc_paper(paper)
    for paper in papers
]

for paper in normalized_papers:
    print(paper)
```

```python
import html


def normalize_europe_pmc_paper(paper):

    # ----- Journal -----
    journal_info = paper.get("journalInfo") or {}
    journal_data = journal_info.get("journal") or {}
    journal_title = journal_data.get("title")

    # ----- Authors -----
    author_string = paper.get("authorString", "")

    authors = [
        author.strip()
        for author in author_string.rstrip(".").split(",")
    ] if author_string else []

    # ----- Title -----
    title = paper.get("title")

    if title:
        title = html.unescape(title)

    return {
        "paper_id": paper.get("id"),
        "title": title,
        "authors": authors,
        "abstract": paper.get("abstractText"),
        "year": paper.get("pubYear"),
        "doi": paper.get("doi"),
        "journal": journal_title,
        "source": "Europe PMC",
        "source_id": paper.get("id")
    }
```
